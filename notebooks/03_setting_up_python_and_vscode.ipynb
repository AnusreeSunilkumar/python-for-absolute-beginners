## 2. Supported Python Version 

 When installing Python for the first time, use the latest stable Python 3 release available from the official Python website. 
 
 Avoid versions marked as: 
 
 - Alpha 
 - Beta 
 - Release candidate 
 - Pre-release 
 
 These versions are intended for testing and may still change. 
 
 > This tutorial uses Python 3. Do not install Python 2.

## 3. Check Whether Python Is Already Installed 

Before installing Python, check whether it is already available on your computer. Open a terminal. 

### Windows 

Open **Command Prompt**, **PowerShell**, or the terminal in Visual Studio Code and run: 

```bash 
python --version 
``` 

You can also try: 

```bash 
py --version 
``` 

### macOS or Linux 

Run: 
```bash 
python3 --version 
``` 

If Python is installed, you should see output similar to: 
```text 
Python 3.14.6
``` 

Your version number may be different. If you see Python 3.12 or a newer stable version, you may continue to the Visual Studio Code installation. If the command is not recognized, install Python using the instructions below.

## 4. Install Python 

Download Python only from the official website: 

[Download Python](https://www.python.org/downloads/)

### Windows Installation

1. Open the official Python download page. 
2. Download the latest stable Python 3 installer for Windows. 
3. Open the downloaded installer. 
4. Select **Add Python to PATH**, if this option is displayed. 
5. Select **Install Now**. 
6. Wait for the installation to finish. 
7. Close and reopen your terminal. 

> **Important:** Adding Python to `PATH` allows you to run Python commands from a terminal.

### macOS Installation 

1. Open the official Python download page. 
2. Download the latest stable installer for macOS. 
3. Open the downloaded installer. 
4. Follow the installation instructions. 
5. Close and reopen your terminal. 

On macOS, you will normally use: 

```bash 
python3 
``` 

instead of: 

```bash 
python 
```

### Linux Installation 

Many Linux distributions already include Python. 

Check the installed version: 

```bash 
python3 --version 
``` 

If Python is not installed, use the package manager provided by your Linux distribution. 

For example, on Ubuntu or Debian-based systems: 

```bash 
sudo apt update 
sudo apt install python3 python3-venv python3-pip 
``` 

Package-management commands may differ between Linux distributions.

> Verify the Python Installation using step 3

## 7. Install the Required Extensions 

Visual Studio Code does not include full Python and notebook support by default. 

You need to install two extensions: 

- **Python**, published by Microsoft 
- **Jupyter**, published by Microsoft

### Installing an Extension 

1. Open Visual Studio Code. 
2. Select the **Extensions** icon in the left sidebar. 
3. Search for `Python`. 
4. Select the extension published by **Microsoft**. 
5. Select **Install**. 
6. Search for `Jupyter`. 
7. Select the extension published by **Microsoft**. 
8. Select **Install**. 

You can also open the Extensions panel using: 

```text 
Ctrl + Shift + X 
``` 

On macOS, use: 

```text 
Command + Shift + X 
``` 

> Check the publisher before installing an extension. The required Python and Jupyter extensions are published by Microsoft.

## 11. What Is a Virtual Environment? 

A virtual environment is an isolated Python environment created for one project. It keeps the packages used by this tutorial separate from packages used by other Python projects. 

We will name the virtual environment: 
```text 
.venv 
``` 

The period at the beginning is part of the name.

## 12. Create a Virtual Environment 

Make sure the terminal is open inside the repository folder. 

### Windows 

Run: 
```bash 
python -m venv .venv 
``` 

If `python` is not recognized, try: 
```bash 
py -m venv .venv 
``` 

### macOS or Linux 

Run: 
```bash 
python3 -m venv .venv
``` 

 After the command finishes, a folder named `.venv` should appear inside the repository. Do not manually edit the files inside `.venv`.

## 13. Activate the Virtual Environment 

Activating the environment makes the terminal use the Python installation and packages inside `.venv`. 

### Windows Command Prompt 
```bat 
.venv\Scripts\activate 
```

### Windows PowerShell 
```powershell 
.venv\Scripts\Activate.ps1 
``` 

### macOS or Linux 
```bash 
source .venv/bin/activate 
``` 

After activation, the terminal may display: 

```text 
(.venv) 
``` 

at the beginning of the command line. 

Example: 

```text 
(.venv) C:\Projects\python-for-absolute-beginners> 
``` 

This indicates that the virtual environment is active.

## 14. Upgrade `pip` 

`pip` is a tool used to install Python packages. 

With the virtual environment active, run: 

### Windows 

```bash 
python -m pip install --upgrade pip 
``` 

### macOS or Linux 

```bash 
python3 -m pip install --upgrade pip 
``` 

Using `python -m pip` helps ensure that packages are installed into the selected Python environment.

## 15. Install the Tutorial Requirements 

The repository contains a file named: 

```text 
requirements.txt 
``` 

It lists the packages needed to run the notebooks. 

With `.venv` active, run: 

### Windows 

```bash 
python -m pip install -r requirements.txt 
``` 

### macOS or Linux 

```bash 
python3 -m pip install -r requirements.txt 
``` 

A minimal `requirements.txt` for this tutorial may contain: 

```text 
jupyterlab>=4.0,<5.0 ipykernel>=6.0,<7.0 
``` 

You do not need to install standard-library modules such as: - `math` - `random` - `pathlib` - `json` - `csv` 

They are included with Python.

## 16. Select the Python Interpreter 

Visual Studio Code needs to know which Python environment should run your programs. 

1. Open the Command Palette: 
```text 
Ctrl + Shift + P 
``` 

On macOS: 
```text 
Command + Shift + P 
``` 

2. Search for: 
```text 
Python: Select Interpreter 
``` 

3. Select the interpreter inside `.venv`. It may be displayed as: 
```text 
Python 3.x.x ('.venv': venv) 
``` 
The exact wording may differ. You may also select the Python version shown in the Visual Studio Code status bar. 

> Select the interpreter located inside the project’s `.venv` folder, not a different global Python installation.

## 17. Select the Notebook Kernel 

A Jupyter notebook uses a **kernel** to run Python code. 

1. Open a notebook from the `notebooks` folder. 
2. Select **Select Kernel** in the top-right corner. 
3. Choose **Python Environments**, if that option appears. 
4. Select the `.venv` environment. 

The selected kernel should be the same environment where you installed the requirements. If the wrong kernel is selected, a package may appear to be missing even though you already installed it.

## 22. Common Setup Problems

### `python` Is Not Recognized

Example:

```text 
'python' is not recognized as an internal or external command 
```

Try: 
```bash 
py --version 
``` 

If `py` works, use `py` to create the virtual environment: 

```bash 
py -m venv .venv 
``` 

If neither command works: 

1. Confirm that Python is installed. 
2. Close and reopen the terminal. 
3. Restart Visual Studio Code. 
4. Reinstall Python and select **Add Python to PATH**, if available.

### The Notebook Cannot Find a Package 

You may see: 

```text 
ModuleNotFoundError 
``` 

Possible causes include: 
- The requirements were installed outside `.venv`. 
- The wrong kernel is selected. 
- The package was not installed. Check the selected kernel and then run: 

### Windows 

```bash 
python -m pip install -r requirements.txt 
``` 

### macOS or Linux 

```bash 
python3 -m pip install -r requirements.txt 
```

### The Terminal Opens in the Wrong Folder 

Use `cd` to enter the repository folder. 

Example: 
```bash 
cd python-for-absolute-beginners 
``` 

You can also close Visual Studio Code and reopen the correct folder using: 
```text 
File → Open Folder 
```

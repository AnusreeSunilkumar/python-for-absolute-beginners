## 2. Understanding the Instruction 

Let us examine each part of the instruction: 
```python 
print("Hello, world!") 
```

### `print`

`print` is the name of a Python function. 

A **function** is a reusable instruction that performs a particular task. 

The task of the `print()` function is to display information. 

You will learn more about functions in a later lesson.

### Parentheses 

The opening and closing parentheses are: 
```text 
( ) 
``` 

The information that we want to display is placed inside the parentheses. 

```python 
print("Hello, world!") 
```

### Quotation Marks 

The quotation marks tell Python that the content inside them is text. 

```text 
"Hello, world!" 
``` 
Text written inside quotation marks is called a **string**. 

You will learn more about strings in a later lesson.

## 5. Quotation Marks Inside Text 

Sometimes the message itself contains quotation marks. 

You can use single quotation marks outside when the text contains double quotation marks: 
```python

``` 

Output: 
```text 
She said, "Hello!" 
``` 

You can use double quotation marks outside when the text contains an apostrophe:

```python 
print("Python is beginner-friendly.") 
print("It's time to learn Python!") 
```

Output: 
```text 
Python is beginner-friendly. 
It's time to learn Python! 
``` 

The opening and closing quotation marks must match. 

Correct: 
```python 
print("Hello") 
``` 

Correct: ```python 
print('Hello') 
``` 

Incorrect: 
```
print("Hello') 
```

````python
print('She said, "Hello!"') 
````

````python
print("Python is beginner-friendly.") 
print("It's time to learn Python!") 
````

## 11. Changing the Separator 

When several items are passed to `print()`, Python normally separates them using a space. 

```python 
print("red", "green", "blue") 
``` 

Output: 
```text 
red green blue 
``` 

You can change the separator using `sep`. 
```python 
print("red", "green", "blue", sep=", ") 
``` 

Output: 
```text red, green, blue 
``` 

Another example: 
```python 
print(2026, 7, 28, sep="-") 
``` 

Output: 
```text 
2026-7-28 
``` 

You do not need to memorize `sep` now. Remember only that it changes what appears between printed items.

````python
print("red", "green", "blue") 
````

````python
print("red", "green", "blue", sep=", ") 
````

````python
print(2026, 7, 28, sep="-") 
````

## 12. Controlling the End of the Output 

Normally, every `print()` instruction ends by moving to a new line. 
```python 
print("Hello") 
print("World") 
``` 
Output: 
```text 
Hello 
World 
``` 

You can change this behavior using `end`. 
```python 
print("Hello", end=" ") 
print("World") 
``` 

Output: 
```text 
Hello World 
``` 

In this example, the first `print()` ends with a space instead of moving to a new line. 

Another example: 
```python 
print("Loading", end="...") 
print("complete") 
``` 

Output: 
```text 
Loading...complete 
``` 

Like `sep`, `end` is optional. 

The normal behavior of `print()` is enough for most beginner examples.

````python
print("Hello") 
print("World") 
````

````python
print("Hello", end=" ") 
print("World") 
````

````python
print("Loading", end="...") 
print("complete") 
````

## 13. Escape Sequences 

An escape sequence is a special combination of characters used inside text. 

Escape sequences begin with a backslash: 
```text 
\ 
``` 

### New Line: `\n` 

The `\n` escape sequence starts a new line. 
```python 
print("First line\nSecond line") 
``` 

Output: 
```text 
First line 
Second line 
``` 

### Tab: `\t` 

The `\t` escape sequence adds horizontal spacing called a tab. 
```python 
print("Name\tScore") 
print("Tom\t90") 
``` 

Output may appear similar to: 

```text 
Name    Score 
Mina    90 
``` 

### Printing a Quotation Mark 

A backslash can be used before a quotation mark: 
```python 
print("He said, \"Hello!\"") 
``` 

Output: 
```text 
He said, "Hello!" 
``` 

However, using different outer quotation marks is often easier: 
```python 
print('He said, "Hello!"') 
```

````python
print("First line\nSecond line") 
````

````python
print("Name\tScore") 
print("Tom\t90") 
````

````python
print("He said, \"Hello!\"") 
````

````python
print('He said, "Hello!"') 
````

## 15. Comments 

A comment is a note written inside the code for people to read. 

Python ignores comments when running the program. 

A comment begins with `#`. 
```python 
# This program displays a greeting 
print("Hello!") 
``` 

Output: 
```text 
Hello! 
``` 

The comment is not displayed. 

You can also place a short comment after an instruction: 

```python 
print("Welcome!") 
# Display a welcome message 
``` 

Use comments to explain **why** something is done when the reason is not clear. Avoid comments that only repeat the code. 

Less useful: 

```python 
# Print hello 
print("Hello") 
``` 

More useful: 

```python 
# Display the title before the lesson instructions 
print("Python Tutorial") 
``` 

You will learn more about writing helpful comments as the programs become larger.

````python
# This program displays a greeting 
print("Hello!") 
````

## 16. Common Mistakes 

Making mistakes is a normal part of programming. 
Read the error message and compare your code with the correct example. 

### Missing Quotation Mark 

Incorrect: 
```python 
print("Hello) 
``` 
Correct: 
```python 
print("Hello") 
``` 

### Missing Closing Parenthesis 

Incorrect: 
```python 
print("Hello" 
``` 
Correct: 
```python 
print("Hello") 
``` 

### Mismatched Quotation Marks 

Incorrect: 
```python 
print("Hello') 
``` 
Correct: 
```python 
print("Hello") 
``` 

### Text Without Quotation Marks 

Incorrect: 
```python 
print(Hello) 
``` 
Correct: 
```python 
print("Hello") 
``` 

Without quotation marks, Python does not recognize `Hello` as text. 

### Using a Capital `P` 

Incorrect: 
```python 
Print("Hello") 
``` 
Correct: 
```python 
print("Hello") 
``` 

Python is case-sensitive. 
This means: 
```text 
print 
``` 
and: 
```text 
Print 
``` 
are considered different names. 

### Adding Quotation Marks Around a Calculation 

This displays the calculation as text: 
```python 
print("5 + 3") 
``` 
Output: 
```text 
5 + 3 
``` 

This performs the calculation: 
```python 
print(5 + 3) 
``` 
Output: 
```text 
8 
```

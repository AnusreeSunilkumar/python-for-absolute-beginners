## 1. What is a Computer Program?

A computer program is a set of instructions that tells computers what to do. 

A program is similar to a recipe. A recipe contains ingredients, a sequence of steps, and a finished result.

We can compare these parts to a program:

- **Input:** The ingredients
- **Processing:** The cooking steps
- **Output:** The finished meal

Similarly, a computer program contains instructions that the computer follows in a specific order. A computer follows the instructions exactly as they are written.

## 2. What is Programming?

Programming is the process of writing the instructions that a computer can understand and execute.

Programming usually involves:

- Understanding a problem
- Breaking the problem into smaller steps
- Writing instructions
- Running the instructions
- Checking the result
- Correcting mistakes
- Improving the solution

Programming is not only about writing code. It is mainly about solving problems logically.

## 3. Computers are very Literal

People can understand incomplete or unclear instructions by using common sense.

Computers cannot do this.

For example, consider the following instruction:

> Make a cup of tea. 

A person may understand what to do. A computer would need more detailed instructions.

### More Detailed Instructions

1. Take a cup.
2. Place a tea bag inside the cup.
3. Fill a kettle with water.
4. Boil the water.
5. Pour the hot water into the cup.
6. Wait for three minutes.
7. Remove the tea bag.

The instructions must be:

- Clear
- Detailed
- Correctly ordered
- Unambiguous

## 4. Input, Processing, and Output

Many programs can be understood using three basic stages:

Input → Processing → Output

### Input

Input is the information given to a program.

Examples include:

- A name entered by a user
- A number typed using the keyboard
- An image captured by a camera
- A file opened by a program

### Processing

Processing is the work performed on the input.

Examples include:

- Adding two numbers
- Comparing two values
- Searching for a word
- Resizing an image

### Output

Output is the result produced by the program.

Examples include:

- Text displayed on the screen
- A calculated total
- A saved file
- A warning message

### Example: Adding Two Numbers

- **Input:** Two numbers
- **Processing:** Add the numbers
- **Output:** The total

## 5. What is an Algorithm?

An algorithm is a clear sequence of steps used to solve a problem.

An algorithm does not have to be written in a programming language.

For example, an algorithm for finding the larger of two numbers could be:

1. Read the first number.
2. Read the second number.
3. Compare the two numbers.
4. Display the larger number.

Algorithms help programmers plan the solution before writing code.

## 6. What is a Programming Language?

A programming language is a language used to write instructions that a computer can execute.

Computers ultimately work with instructions represented using binary digits:

```0 and 1```

A single binary digit, either 0 or 1, is called a bit.

Groups of bits can represent:

- Numbers
- Text
- Images
- Sound
- Program instructions

Writing an entire program directly using binary instructions would be extremely difficult for humans. Programming languages allow programmers to express instructions in a more understandable form.

Programming languages are often described according to how close they are to the computer's hardware.

### Machine Language

Machine language contains instructions that a computer's processor can execute directly.

These instructions are represented as patterns of binary digits.

A simplified example might look like this:

`10110000 01100001`

### Assembly Language

Assembly language provides short, readable names for low-level machine operations.

A simplified assembly instruction might look like this:

`ADD R1, R2`

Assembly language is easier for humans to read than machine language.

An assembly-language program must be translated into machine language by a program called an assembler.

Machine language and assembly language are commonly called low-level languages.

### High-Level Programming Languages

High-level programming languages allow programmers to write instructions using syntax that is easier for humans to read and understand. Different programming languages have different rules and uses. 

High level languages also need to be translated into a form that the computer can execute. Different languages use different translation methods.

Some examples of programming languages are:

- Python
- Java
- JavaScript
- C#



## 7. More Programming Concepts

### Source Code

Source code is the set of instructions written by a programmer. Source code is usually saved inside files.

### Syntax

Syntax refers to the rules for writing valid code in a programming language. It is similar to grammar in a spoken language.

If the code does not follow these rules, the computer may not be able to run it.

You do not need to memorize every syntax rule immediately. You will learn the rules gradually by reading, writing, and running code.

### Types of Errors

An error is a problem that prevents a program from working as expected. There are several common types of errors.

1. Syntax Error

A syntax error happens when the code does not follow the language rules.

2. Runtime Error

A runtime error happens while the program is running.

3. Logic Error

A logic error happens when the program runs but produces an incorrect result.

### Bugs and Debugging

A bug is a mistake or problem in a program.

Debugging is the process of finding and correcting bugs.

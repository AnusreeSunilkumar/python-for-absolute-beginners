# Jupyter Notebooks

Jupyter notebooks allow you to combine explanations, code, and output in one interactive document.

A notebook file uses the `.ipynb` extension.

## Different Types of Cells

A Jupyter notebook mainly contains two types of cells:

### Markdown Cells

Markdown cells are used to write explanations, instructions, headings, notes, and formatted text.

They do not run Python code or produce program output.

Markdown cells can contain:

- Headings
- Paragraphs
- Lists
- Links
- Images
- Tables
- Formatted code examples

This text is written inside a Markdown cell.

### Code Cells

Code cells are used to write and run Python code.

A code cell may or may not produce output.

A code cell may display output:

```python
print("Hello!")
```

Some code cells run without displaying anything:

```python
name = "Anu"
```

The second cell stores a value, but it does not display it.

## Running Notebook Cells

Select a cell and press:

- `Shift + Enter` to run the cell and move to the next one
- `Ctrl + Enter` to run the cell and remain on it

`Run all` option on top of the notebook will run all cells in sequence.

You can also use the **Run** button near the cell.

### Run Cells in Order

Notebook cells can remember values from previously executed cells.

For best results, run the notebook from top to bottom.

### Editing and Rerunning a Cell

Click inside a cell to edit its contents.

After changing a code cell, run the cell again to execute the updated code. Editing a cell does not automatically update its previous output.

### Clearing Cell Output

You can remove displayed output without deleting the code.

Use the notebook option to clear the output of one cell or all cells.

Clearing output does not remove the code itself.

### Stopping a Running Cell

Some code may take a long time to finish.

A loop may also run forever if its stopping condition is incorrect.

You can stop the current execution using the Interrupt or Stop button.

### Saving the Notebook

Save the notebook regularly.

Use:

`Ctrl + S` on Windows or Linux
`Command + S` on macOS

## The Kernel

The kernel is the process that runs the Python code in the notebook.

It remembers:

- Variables
- Imported modules
- Functions
- Results from previously executed cells

The kernel continues to remember these values until it is restarted or stopped.

### Selecting the Kernel

Before running a notebook, check that the Python environment created for this tutorial is selected as the kernel. Kernel selection is explained in the setup notebook.


### Restarting the Kernel

Restart the kernel when:

- Variables contain unexpected values
- Cells were executed out of order
- The notebook behaves differently than expected

After restarting, run all cells again from the beginning.
